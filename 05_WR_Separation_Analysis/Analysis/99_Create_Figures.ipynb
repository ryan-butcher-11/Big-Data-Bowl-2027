```python
from PIL import Image, ImageDraw, ImageFont

image_path = str(SEPARATION_IMAGES_DIR / "wr_decel_vs_cut_sep.png")
output_path = image_path.replace(".png", "_fig2.png")

# Load image
img = Image.open(image_path).convert("RGB")

# Add header space
header_height = 70
canvas = Image.new("RGB", (img.width, img.height + header_height), "white")
canvas.paste(img, (0, header_height))

# Add centered, bold caption
draw = ImageDraw.Draw(canvas)
font = ImageFont.truetype(
    "/System/Library/Fonts/HelveticaNeue.ttc",
    24,
    index=1
)

caption = "Figure 2"
bbox = draw.textbbox((0, 0), caption, font=font)

text_width = bbox[2] - bbox[0]
text_height = bbox[3] - bbox[1]

x = (canvas.width - text_width) // 2
y = (header_height - text_height) // 2 - bbox[1] + 40

draw.text((x, y), caption, fill="black", font=font)

# Save
canvas.save(output_path)
```

## WR Summary Tables
***

```python
from PIL import Image, ImageDraw, ImageFont

# ============================================
# File Paths
# ============================================

image_dir = SEPARATION_IMAGES_DIR

accel_path = f"{image_dir}/accel_top5.png"
decel_path = f"{image_dir}/decel_top5.png"
output_path = f"{image_dir}/accel_decel_top5.png"


# ============================================
# Load Images
# ============================================

accel_img = Image.open(accel_path).convert("RGB")
decel_img = Image.open(decel_path).convert("RGB")


# ============================================
# Resize to Same Height
# ============================================

target_height = max(accel_img.height, decel_img.height)

def resize_to_height(img, height):
    new_width = int(img.width * height / img.height)
    return img.resize((new_width, height), Image.Resampling.LANCZOS)

accel_img = resize_to_height(accel_img, target_height)
decel_img = resize_to_height(decel_img, target_height)


# ============================================
# Combine Images with Header Space
# ============================================

gap = 20
header_height = 100

combined_width = accel_img.width + gap + decel_img.width
combined_height = target_height + header_height

combined_img = Image.new(
    "RGB",
    (combined_width, combined_height),
    color="white"
)

# Paste images below header
combined_img.paste(accel_img, (0, header_height))
combined_img.paste(decel_img, (accel_img.width + gap, header_height))


# ============================================
# Add Centered Bold Caption
# ============================================

draw = ImageDraw.Draw(combined_img)

font = ImageFont.truetype(
    "/System/Library/Fonts/HelveticaNeue.ttc",
    72,
    index=1
)

caption = "Figure 6 - Top WR Prospects Based On Acceleration And Deceleration"

bbox = draw.textbbox((0, 0), caption, font=font)
text_width = bbox[2] - bbox[0]
text_height = bbox[3] - bbox[1]

text_x = (combined_width - text_width) // 2
text_y = (header_height - text_height) // 2 - bbox[1]

draw.text(
    (text_x, text_y),
    caption,
    fill="black",
    font=font
)


# ============================================
# Save
# ============================================

combined_img.save(output_path)

print(f"Saved combined image to: {output_path}")
```
